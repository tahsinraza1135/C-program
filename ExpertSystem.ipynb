```python
# 1. Knowledge Base

# A student is represented as a frame containing attributes.

def create_student(name, marks, entrance_exam, documents_verified):
  return {
"name": name,
"marks": marks,
"entrance_exam": entrance_exam,
"documents_verified": documents_verified
}

# 2. Inference Engine

# Applies rules to the facts stored in the knowledge base.

def make_decision(student):
  name = student["name"]
  marks = student["marks"]
  entrance = student["entrance_exam"]
  documents_verified = student["documents_verified"]

  reasons = []

# Rule 1: Validate academic marks
  if marks is None:
    reasons.append("Academic marks are missing.")

  elif marks < 0 or marks > 100:
    return "Invalid Data", [
        "Marks must be between 0 and 100."
    ]

  elif marks < 60:
    return "Not Eligible", [
        f"Marks are {marks}%, below the required 60%."
    ]

# Rule 2: Check entrance examination
  if entrance is None:
    reasons.append("Entrance exam status is unknown.")

  elif entrance is False:
    return "Not Eligible", [
        "The entrance examination was not passed."
    ]

# Rule 3: Verify required documents
  if documents_verified is None:
    reasons.append("Document verification status is unknown.")

  elif documents_verified is False:
    reasons.append("Required documents are not verified.")

# Rule 4: Make the final decision
  if reasons:
    return "Conditional Review", reasons

  return "Eligible", [
    "Academic marks meet the minimum requirement.",
    "Entrance examination is passed.",
    "Required documents are verified."
]

# 3. Test the Knowledge Base with Different Students

students = [
create_student("Aman", 85, True, True),
create_student("Simran", 52, True, True),
create_student("Ravi", 75, False, True),
create_student("Neha", 78, True, None),
create_student("Karan", 70, True, False)
]

# 4. Display Decisions and Explanations

for student in students:
  decision, reasons = make_decision(student)

  print("\nStudent:", student["name"])
  print("Admission Decision:", decision)
  print("Reasoning:")

  for reason in reasons:
    print("-", reason)

```
